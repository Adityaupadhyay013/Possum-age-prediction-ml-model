```python
L1 = df['hdlngth'].mean() - 3*df['hdlngth'].std()
U1 = df['hdlngth'].mean() + 3*df['hdlngth'].std()
print(L1 , U1)
```

```python
L2 = df['skullw'].quantile(0.75)-1.5*(df['skullw'].quantile(0.75)-df['skullw'].quantile(0.25))
U2 = df['skullw'].quantile(0.75)+1.5*(df['skullw'].quantile(0.75)-df['skullw'].quantile(0.25))
print(L2 , U2)
df['skullw'].describe()
```

```python
L3 = df['taill'].mean() - 3*df['taill'].std()
U3 = df['taill'].mean() + 3*df['taill'].std()
print(L3 , U3)
df['taill'].describe()
```

```python
L4 = df['eye'].mean() - 3*df['eye'].std()
U4 = df['eye'].mean() + 3*df['eye'].std()
print(L4 , U4)
df['eye'].describe()
```

```python
L5 = df['chest'].mean() - 3*df['chest'].std()
U5 = df['chest'].mean() + 3*df['chest'].std()
print(L5 , U5)
df['chest'].describe()
```

```python
L6 = df['belly'].mean() - 3*df['belly'].std()
U6 = df['belly'].mean() + 3*df['belly'].std()
print(L6 , U6)
df['belly'].describe()
```

```python
def Outliers_Treatment(df):
    df['hdlngth'] = df['hdlngth'].clip(lower = L1 , upper = U1)
    df['skullw'] = df['skullw'].clip(lower = L2 , upper = U2)
    df['taill'] = df['taill'].clip(lower = L3 , upper = U3)
    df['eye'] = df['eye'].clip(lower = L4 , upper = U4)
    df['chest'] = df['chest'].clip(lower = L5 , upper = U5)
    df['belly'] = df['belly'].clip(lower = L6 , upper = U6)
    df['totlngth'] = df['totlngth'].clip(lower = L_ , upper = U_)
    df['footlgth'] = df['footlgth'].clip(lower = L7 , upper = U7)
    df['earconch'] = df['earconch'].clip(lower = L8 , upper = U8)  
    return df
```

```python
df = Outliers_Treatment(df)
```
